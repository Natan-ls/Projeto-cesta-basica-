Salvar os estabelecimentos em csv

```python
df_mercado = pd.DataFrame(constantes.NOMES_MERCADOS, columns = ['Estabelecimento'])
df_mercado
```

```python
df_posto = pd.DataFrame(constantes.NOMES_POSTOS, columns = ['Estabelecimento'])
df_posto
```

```python
df_mercado['tipo_estabelecimento'] = 'SUPERMERCADO'
df_posto['tipo_estabelecimento'] = 'POSTO_COMBUSTIVEL'
```

```python
df_estabelecimento = pd.concat([df_mercado, df_posto], ignore_index=True)
df_estabelecimento
```

```python
df_estabelecimento.sort_values(by='Estabelecimento')
df_estabelecimento
```

```python
df_estabelecimento.to_csv('db/estabelecimentos.csv', index=False)
```
